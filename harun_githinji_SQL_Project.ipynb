Using aggregate functions

Reecent Sales >=2020

```python
%%sql

SELECT
  orderdate,
  quantity * netprice * exchangerate  AS Net_revenue
FROM
  Sales
WHERE
  orderdate >= '2020-01-01'
Order by
  net_revenue asc
LIMIT 10
```

USING LEFT JOIN IN THE QUERIES

```python
%%sql

SELECT
  s.orderdate,
  p.productname,
  CONCAT(c.givenname, ' ', c.surname) AS Customer_name,
  s.quantity * s.netprice * s.exchangerate  AS Net_revenue
FROM
  Sales s
LEFT JOIN customer c ON s.customerkey = c.customerkey
LEFT JOIN product p ON s.productkey = p.productkey
WHERE
  orderdate >= '2020-01-01'
Order by
  net_revenue asc
LIMIT 10
```

```python
%%sql

SELECT
  s.orderdate,
  p.productname,
  CONCAT(c.givenname, ' ', c.surname) AS Customer_name,
  s.quantity * s.netprice * s.exchangerate  AS Net_revenue,
  CASE WHEN s.quantity * s.netprice * s.exchangerate > 1000 THEN 'High Revenue'
  ELSE 'Low Revenue' END AS Revenue_Category
FROM
  Sales s
LEFT JOIN customer c ON s.customerkey = c.customerkey
LEFT JOIN product p ON s.productkey = p.productkey
WHERE
  s.orderdate::date>='2020-01-01'
Order by
  net_revenue asc
LIMIT 10
```

COUNT (DISTINCT)

Total customers per day in 2023

```python
%%sql

SELECT
  s.orderdate,
  CONCAT(c.givenname,' ',c.surname) AS Customer_name,
  COUNT(DISTINCT s.customerkey) AS Customer_count
FROM
  Sales s
LEFT JOIN customer c ON s.customerkey = c.customerkey
WHERE
  orderdate >= '2023-01-01'
Group by
  orderdate,
  Customer_name
LIMIT 10
```

Daily Customers by Region

```python
%%sql

SELECT DISTINCT
  continent
FROM
  customer
```

```python
%%sql

SELECT
  s.orderdate,
  CONCAT(c.givenname,' ',c.surname) AS Customer_name,
  COUNT(DISTINCT CASE WHEN c.continent = 'Europe' THEN s.customerkey END) AS Europe_count,
  COUNT(DISTINCT CASE WHEN c.continent = 'North America' THEN s.customerkey END) AS North_America_count,
  COUNT(DISTINCT CASE WHEN c.continent = 'Australia' THEN s.customerkey END) AS Australia_count
FROM
  Sales s
LEFT JOIN customer c ON s.customerkey = c.customerkey
Group by
  orderdate,
  Customer_name
LIMIT 10
```

Total net revenue by Category

```python
%%sql

SELECT
  p.categoryname,
  SUM(s.quantity * s.netprice * s.exchangerate) AS Total_revenue,
  SUM(CASE WHEN s.orderdate BETWEEN '2022-01-01' AND '2022-12-31'
    THEN s.quantity * s.netprice * s.exchangerate ELSE NULL END) AS Revenue_2022,
  SUM(CASE WHEN s.orderdate BETWEEN '2023-01-01' AND '2023-12-31'
    THEN s.quantity * s.netprice * s.exchangerate ELSE NULL END) AS Revenue_2023
FROM
  Sales s
LEFT JOIN product p ON s.productkey = p.productkey
Group by
  p.categoryname
```

Statistical Functions

```python
%%sql

SELECT
  orderdate,
  PERCENTILE (0.5) WITHIN GROUP (ORDER BY netprice) as median_price
FROM
  sales
Group by
  orderdate
```
